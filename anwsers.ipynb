```python
#Challenge 3
import pandas as pd

dirty_clients=pd.read_csv('./clients_dirty.csv')

valeur_manquantes=dirty_clients.isna().sum()
print(valeur_manquantes)
print("--------------------------------------------------")
duplicated=dirty_clients.duplicated()
print(duplicated)
print("--------------------------------------------------")
dirty_clients = dirty_clients.drop_duplicates()
print("--------------------------------------------------")
dirty_clients['nom']=dirty_clients['nom'].str.strip()
print("--------------------------------------------------")
dirty_clients['ville']=dirty_clients['ville'].str.strip()
print("--------------------------------------------------")
dirty_clients['ville']=dirty_clients['ville'].str.lower()
print("--------------------------------------------------")
dirty_clients["age"]=dirty_clients["age"].replace('N/A',pd.NA)
print("--------------------------------------------------")
dirty_clients['age']=pd.to_numeric(dirty_clients['age'],errors='coerce')
print("--------------------------------------------------")
dirty_clients["salaire"]=dirty_clients["salaire"].astype(str).str.replace("DH","",regex=False).str.strip()
print("--------------------------------------------------")
dirty_clients["salaire"]=pd.to_numeric(dirty_clients["salaire"], errors='coerce')
print("--------------------------------------------------")
salaire_neg=dirty_clients[dirty_clients["salaire"]<0]
print(salaire_neg)
print("--------------------------------------------------")
dirty_clients.loc[dirty_clients["salaire"]<0,"salaire"]=pd.NA
dirty_clients["salaire"]=dirty_clients["salaire"].fillna(dirty_clients["salaire"].median())
print("--------------------------------------------------")
dirty_clients['date_inscription']=pd.to_datetime(dirty_clients["date_inscription"])
print("--------------------------------------------------")
print(dirty_clients.dtypes)
print("--------------------------------------------------")
print(len(dirty_clients))
print("--------------------------------------------------")
print(dirty_clients.describe())
print("--------------------------------------------------")

```

```python
#Challenge 5

import pandas as pd

ventes=pd.read_csv('./ventes.csv')

ca_total=(ventes["prix"]*ventes["quantite"]).sum()
print(ca_total)

ca_moyen=(ventes["prix"]*ventes["quantite"]).mean()
print(ca_moyen)

ca_max = (ventes["prix"] * ventes["quantite"]).max()
print(ca_max)

ca_min=(ventes["prix"]*ventes["quantite"]).min()
print(ca_min)

median=(ventes["prix"]*ventes["quantite"]).median()
print(median)

ecart_types=(ventes["prix"]*ventes["quantite"]).std()
print(ecart_types)

variance=(ventes["prix"]*ventes["quantite"]).var()
print(variance)

quantiles=(ventes["prix"]*ventes["quantite"]).quantile([0.1, 0.25, 0.5, 0.75, 0.9])
print(quantiles)
```

```python
# Challenge 5
# Analyse par produit
import pandas as pd

ventes=pd.read_csv('./ventes.csv')

qte_par_produit=ventes.groupby('produit')['quantite'].sum()
print(qte_par_produit)

ventes['chiffre_affaires'] = ventes['prix'] * ventes['quantite']
ca_total= ventes.groupby("produit")['chiffre_affaires'].sum()
print(ca_total)

produit_top = ca_total.idxmax()
print(produit_top)
```

```python
# Challenge 5
# Analyse avancée

# 15. Utiliser `agg()` pour calculer plusieurs statistiques simultanément

import pandas as pd

ventes=pd.read_csv('./ventes.csv')

ventes["chiffre_affaires"]= ventes["prix"] * ventes["quantite"]

stats= ventes.groupby("produit")["chiffre_affaires"].agg(["sum", "mean", "max"])
print(stats)
```

```python
#Challenge 6


# 10. Utiliser `apply()` pour réaliser au moins une transformation personnalisée.

import pandas as pd

employes=pd.read_csv('./employes.csv')

employes['salaire_moyen_departement']=employes.groupby("departement")["salaire"].transform('mean')

employes['ecart_moyenne'] = employes['salaire'] - employes['salaire_moyen_departement']

employes['moyen']=employes.groupby("departement")["performance"].transform('mean')

filter_salaire=employes['salaire']>employes['salaire_moyen_departement']
employes_au_dessus = employes[filter_salaire]
print(employes_au_dessus)


stats = employes.groupby("departement")[["salaire", "performance"]].agg(["sum", "mean", "max"])
print(stats)

def categorie_perf(p):
    if p >= 90:
        return 'Excellent'
    elif p >= 80:
        return 'Bon'
    else:
        return 'Moyen'

employes['categorie_performance'] = employes['performance'].apply(categorie_perf)
```
