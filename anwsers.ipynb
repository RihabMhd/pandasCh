```python
#Challenge 1
import pandas as pd

clients=pd.read_csv('./clients.csv')
print(clients)
print('-------------------------------------------------------------------------')
premieres_lignes=clients.head()
dernieres_lignes=clients.tail()

print("Premiere lignes :",premieres_lignes)
print('----------------------------------------------------------------------------')
print("Derniere lignes :",dernieres_lignes)
print('----------------------------------------------------------------------------')
nmbr_ligne_colonne=clients.shape[1]
print("Nombre des colonnes :",nmbr_ligne_colonne)
print('----------------------------------------------------------------------------')
nom_colonne=clients.columns
print(list(nom_colonne))
print('----------------------------------------------------------------------------')
indexes=clients.index
print("Index :",indexes)
print('----------------------------------------------------------------------------')
file_type=clients.dtypes
print("File type :",file_type)
print('----------------------------------------------------------------------------')
clients.info()
print('----------------------------------------------------------------------------')
description=clients.describe()
descriptioncatego=clients.describe(include="object")
print("Description :",description)
print('----------------------------------------------------------------------------')
colonnes_numerique=clients.select_dtypes(include="number").columns
colonnes_catego=clients.select_dtypes(include="object").columns
print("Colonne categor :",colonnes_catego)
print("Colonne numerique :",colonnes_numerique)
print('----------------------------------------------------------------------------')
ville_unique=clients['ville'].unique()
print("Ville unique :",ville_unique)
print('----------------------------------------------------------------------------')
nombre_ville_unique=clients['ville'].nunique()
print("Nombre Ville unique :",nombre_ville_unique)
print('----------------------------------------------------------------------------')
nombre_ville_par_clients=clients['ville'].value_counts()
print("Nombre ville par clients :",nombre_ville_par_clients)
print('----------------------------------------------------------------------------')
valeur_null_colonne=clients.isnull().sum()
valeur_null_total=clients.isnull().sum().sum()
print("Valeur null dans les colonnes :",valeur_null_colonne)
print("Valeur null total",valeur_null_total)

```

```python
#Challenge2
import pandas as pd

clients=pd.read_csv('./clients.csv')

only_names=clients["nom"]
print(only_names)
print("--------------------------------------------------")
names_ages_cities=clients[["nom","age","ville"]]
print(names_ages_cities)
print("--------------------------------------------------")
print(clients.iloc[0:3])
print("--------------------------------------------------")
print(clients[clients['age']>30])
print("--------------------------------------------------")
print(clients[clients['salaire']>6000])
print("--------------------------------------------------")
print(clients[clients['ville']=="Casablanca"])
print("--------------------------------------------------")
print(clients[(clients['sexe']=="F") & (clients['age']>30)])
print("--------------------------------------------------")
print(clients[(clients['ville']=="Casablanca") | (clients['ville']=="Rabat")])
print("--------------------------------------------------")
print(clients[clients['ville'].isin(["Casablanca","Rabat","Tanger"])])
print("--------------------------------------------------")
print(clients[clients['age'].between(25,35)])
print("--------------------------------------------------")
print(clients[~(clients['ville']=="Casablanca")])
print("--------------------------------------------------")
print(clients.loc[0, 'nom'])# index label, nom colonne
print(clients.iloc[0, 1]) # position ligne, position colonne
```
