```python
#Challenge 1
import pandas as pd

clients=pd.read_csv('./clients.csv')
print(clients)
print('-------------------------------------------------------------------------')
premieres_lignes=clients.head()
dernieres_lignes=clients.tail()

print("Premiere lignes :",premieres_lignes)
print('----------------------------------------------------------------------------')
print("Derniere lignes :",dernieres_lignes)
print('----------------------------------------------------------------------------')
nmbr_ligne_colonne=clients.shape[1]
print("Nombre des colonnes :",nmbr_ligne_colonne)
print('----------------------------------------------------------------------------')
nom_colonne=clients.columns
print(list(nom_colonne))
print('----------------------------------------------------------------------------')
indexes=clients.index
print("Index :",indexes)
print('----------------------------------------------------------------------------')
file_type=clients.dtypes
print("File type :",file_type)
print('----------------------------------------------------------------------------')
clients.info()
print('----------------------------------------------------------------------------')
description=clients.describe()
descriptioncatego=clients.describe(include="object")
print("Description :",description)
print('----------------------------------------------------------------------------')
colonnes_numerique=clients.select_dtypes(include="number").columns
colonnes_catego=clients.select_dtypes(include="object").columns
print("Colonne categor :",colonnes_catego)
print("Colonne numerique :",colonnes_numerique)
print('----------------------------------------------------------------------------')
ville_unique=clients['ville'].unique()
print("Ville unique :",ville_unique)
print('----------------------------------------------------------------------------')
nombre_ville_unique=clients['ville'].nunique()
print("Nombre Ville unique :",nombre_ville_unique)
print('----------------------------------------------------------------------------')
nombre_ville_par_clients=clients['ville'].value_counts()
print("Nombre ville par clients :",nombre_ville_par_clients)
print('----------------------------------------------------------------------------')
valeur_null_colonne=clients.isnull().sum()
valeur_null_total=clients.isnull().sum().sum()
print("Valeur null dans les colonnes :",valeur_null_colonne)
print("Valeur null total",valeur_null_total)

```

```python
#Challenge 3
import pandas as pd

dirty_clients=pd.read_csv('./clients_dirty.csv')

valeur_manquantes=dirty_clients.isna().sum()
print(valeur_manquantes)
print("--------------------------------------------------")
duplicated=dirty_clients.duplicated()
print(duplicated)
print("--------------------------------------------------")
dirty_clients = dirty_clients.drop_duplicates()
print("--------------------------------------------------")
dirty_clients['nom']=dirty_clients['nom'].str.strip()
print("--------------------------------------------------")
dirty_clients['ville']=dirty_clients['ville'].str.strip()
print("--------------------------------------------------")
dirty_clients['ville']=dirty_clients['ville'].str.lower()
print("--------------------------------------------------")
dirty_clients["age"]=dirty_clients["age"].replace('N/A',pd.NA)
print("--------------------------------------------------")
dirty_clients['age']=pd.to_numeric(dirty_clients['age'],errors='coerce')
print("--------------------------------------------------")
dirty_clients["salaire"]=dirty_clients["salaire"].astype(str).str.replace("DH","",regex=False).str.strip()
print("--------------------------------------------------")
dirty_clients["salaire"]=pd.to_numeric(dirty_clients["salaire"], errors='coerce')
print("--------------------------------------------------")
salaire_neg=dirty_clients[dirty_clients["salaire"]<0]
print(salaire_neg)
print("--------------------------------------------------")
dirty_clients.loc[dirty_clients["salaire"]<0,"salaire"]=pd.NA
dirty_clients["salaire"]=dirty_clients["salaire"].fillna(dirty_clients["salaire"].median())
print("--------------------------------------------------")
dirty_clients['date_inscription']=pd.to_datetime(dirty_clients["date_inscription"])
print("--------------------------------------------------")
print(dirty_clients.dtypes)
print("--------------------------------------------------")
print(len(dirty_clients))
print("--------------------------------------------------")
print(dirty_clients.describe())
print("--------------------------------------------------")

```
